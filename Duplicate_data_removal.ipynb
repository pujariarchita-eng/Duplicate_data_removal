```python

# ============================================================
# AI/ML Internship Task: Duplicate Data Removal Using Pandas
# Company: Veda Technology
# ============================================================

import pandas as pd

# 1. Create a sample dataset containing duplicate records
data = {
    "Student_ID": [101, 102, 103, 104, 102, 105, 103, 106, 101, 107],
    "Name": [
        "Amit", "Priya", "Rahul", "Sneha", "Priya",
        "Neha", "Rahul", "Karan", "Amit", "Riya"
    ],
    "Age": [21, 22, 20, 23, 22, 21, 20, 24, 21, 22],
    "Department": [
        "AI", "Data Science", "Python", "AI", "Data Science",
        "Python", "Python", "AI", "AI", "Data Science"
    ]
}

df = pd.DataFrame(data)

# 2. Display the original dataset
print("=" * 60)
print("ORIGINAL DATASET")
print("=" * 60)
print(df.to_string(index=False))

# 3. Store the original number of rows
original_rows = len(df)
print("\nOriginal number of rows:", original_rows)

# 4. Identify duplicate records
duplicate_mask = df.duplicated()
duplicate_count = duplicate_mask.sum()

print("\nNumber of duplicate records:", duplicate_count)

print("\nDuplicate records identified:")
print(df[duplicate_mask].to_string(index=False))

# 5. Remove duplicate records
cleaned_df = df.drop_duplicates().copy()

# 6. Count rows after cleaning
cleaned_rows = len(cleaned_df)
removed_rows = original_rows - cleaned_rows

# 7. Display the cleaned dataset
print("\n" + "=" * 60)
print("CLEANED DATASET")
print("=" * 60)
print(cleaned_df.to_string(index=False))

# 8. Compare before-and-after row counts
print("\n" + "=" * 60)
print("BEFORE-AND-AFTER COMPARISON")
print("=" * 60)
print("Rows before cleaning:", original_rows)
print("Rows after cleaning: ", cleaned_rows)
print("Duplicate rows removed:", removed_rows)

# 9. Verify that no full-row duplicates remain
print("\nRemaining duplicate records:", cleaned_df.duplicated().sum())

print("\nTask completed successfully!")
```
